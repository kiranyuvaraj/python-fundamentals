### 8. Given the list [21, 44, 22, 878, 55, 90, 17, 68, 69, 91], write a Python program to create two separate lists: one for even numbers and one for odd numbers.

```python
list_numbers=[21,44,22,878,55,90,17,68,69,91]
even_list=[]
odd_list=[]
for i in list_numbers:
  if i%2==0:
    even_list.append(i)
  else:
    odd_list.append(i)
print('Even list',even_list)
print('Odd list',odd_list)
```

### 9. Write a Python program to find the maximum and minimum values in the list: [21, 44, 767, 98, 37].

```python
numbers=[21,44,767,98,37]
max_value=numbers[0]
min_value=numbers[0]
for num in numbers:
  if num>max_value:
    max_value=num
  elif num<min_value:
    min_value=num
print('Minimum Number in List is',min_value)
print('Maximum Number in List is',max_value)
```

### 10. Find the length of the longest word in the string: ‘Data Science and Machine Learning’.

### 18. Given the list [12, 45, 3, 67, 89, 23, 1, 56], write a Python program to find the second largest number in the list.

```python
numbers=[12,45,3,67,89,23,1,56]
largest=numbers[0]
second_largest=numbers[0]
for num in numbers:
  if num>largest:
    second_largest=largest
    largest=num
  elif num>second_largest and num!=largest:
    second_largest=num

print(f"The Second Largest number is: {second_largest}")
```

### 19. Write a Python program to count the number of vowels in the string: "machine learning and artificial intelligence".

```python
str="machine learning and artificial intelligence"
vowels=['a','e','i','o','u']
count=0
for char in str:
  if char in vowels:
    count=count+1
print(f"Number of vowels in the string is: {count}")
```

### 20. Write a Python program to check whether a given string is a palindrome or not.

```python
str=input("Enter a string to check whether it is a palindrome or not: ").lower()
reverse_str=""
for i in range(len(str)-1,-1,-1):
  reverse_str=reverse_str+str[i]
# Add this check to finish the program
if str == reverse_str:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```
